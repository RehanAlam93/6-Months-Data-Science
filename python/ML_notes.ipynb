#### Countplot :
A Countplot is a type of graph used in data science (provided by the Seaborn library) to display the count (frequency) of each category in a categorical variable. It helps us understand how many times each category appears in a dataset.
Example:
Suppose you have a dataset of students with their gender:

* gender

Female

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load sample dataset
tips = sns.load_dataset("tips")

# Create countplot
sns.countplot(x="day", data=tips)

plt.show()
```

#### boxplot :
A Boxplot (Box-and-Whisker Plot) is a graph used in data science (provided by the Seaborn library) to summarize the distribution of numerical data. It shows the minimum value, first quartile (Q1), median (Q2), third quartile (Q3), maximum value, and helps identify outliers.

Example:    
Suppose you have the marks of students: 

40, 45, 50, 55, 60, 65, 70, 75, 95  

A boxplot will show:    

Minimum: 40     
Q1 (25%): 47.5  
Median (50%): 60    
Q3 (75%): 72.5  
Maximum: 95 
If a value is unusually far from the rest (e.g., 95), it may appear as an outlier.  

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load sample dataset
tips = sns.load_dataset("tips")

# Create boxplot
sns.boxplot(x="day", y="total_bill", data=tips)

plt.show()
```

#### heatmap : 
A Heatmap is a graphical representation of data where different colors represent different values. In data science (using the Seaborn library), it is commonly used to visualize correlation between variables or display data in a matrix format.  

A heatmap uses colors to represent these values:    

Dark color → High correlation (closer to 1)     
Light color → Low correlation (closer to 0) 

This makes it easy to identify strong and weak relationships between variables. 

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load sample dataset
tips = sns.load_dataset("tips")

# Calculate correlation matrix
corr = tips.corr(numeric_only=True)

# Create heatmap
sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.show()
```

#### Correlation (corr()) :
Correlation is a statistical measure that shows the strength and direction of the relationship between two numerical variables. In Python, the corr() function (Pandas) is used to calculate the correlation coefficient between columns in a dataset.

Correlation Values:
* +1 → Perfect positive correlation (both variables increase together)
* 0 → No correlation (no relationship)
* -1 → Perfect negative correlation (one variable increases while the other decreases)

```python
import pandas as pd

# Create sample data
data = {
    "Hours": [2, 4, 6, 8],
    "Marks": [40, 55, 70, 85]
}

df = pd.DataFrame(data)

# Find correlation
print(df.corr())
```

#### Scikit-learn (sklearn) :

#### Test (Model Testing) :
Testing is the process of checking how well a trained machine learning model performs on new, unseen data. The model uses X_test to make predictions, and the predicted values are compared with y_test to measure its accuracy.    

Example:    
Suppose a model is trained to predict students' marks based on study hours. 

X_test = [[10]] (10 study hours)

#### Accuracy :
Accuracy is a performance metric that measures how many predictions made by a machine learning model are correct. It is mostly used for classification problems.    

formula 	

		Accuracy= Total Number of Predictions / Number of Correct Predictions

	​


Example:    
Suppose a model predicts whether an email is Spam or Not Spam.  

Total emails = 10   
Correct predictions = 9     
Incorrect predictions = 1   

    Accuracy =  9 /10 = 0.90 = 90% 

This means the model correctly predicted 90% of the emails.

```python
from sklearn.metrics import accuracy_score

# Actual values
y_test = [1, 0, 1, 1, 0]

# Predicted values
y_pred = [1, 0, 1, 0, 0]

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

#### SVM (Support Vector Machine) :
SVM (Support Vector Machine) is a supervised machine learning algorithm used for classification and regression problems. It works by finding the best boundary (called a hyperplane) that separates different classes with the maximum possible margin. 

Example:    
Suppose you want to classify emails as Spam or Not Spam.    

* Input: Email features
* Output: Spam / Not Spam

An SVM model learns from the training data and finds the best boundary to separate spam emails from non-spam emails. It then classifies new emails based on this boundary.

```python
from sklearn.svm import SVC

# Training data
X_train = [[1], [2], [3], [4]]
y_train = [0, 0, 1, 1]

# Create SVM model
svm = SVC()

# Train the model
svm.fit(X_train, y_train)

# Make prediction
prediction = svm.predict([[2.5]])

print(prediction)
```

#### score() :
score() is a method in Scikit-learn used to evaluate the performance (accuracy) of a trained machine learning model on test data.   
* For Regression models (e.g., LinearRegression, SVR), score() returns the R² (R-squared) score.
* For Classification models (e.g., SVC), score() returns the accuracy of the model.

Syntax:     
model.score(X_test, y_test) 

* X_test → Testing input data
* y_test → Actual testing output

Example (Classification):   
Suppose a model predicts whether a student Passes or Fails.

```python
from sklearn.svm import SVC

X_train = [[35], [45], [60], [80]]
y_train = [0, 0, 1, 1]

X_test = [[50], [75]]
y_test = [0, 1]

model = SVC()
model.fit(X_train, y_train)

accuracy = model.score(X_test, y_test)

print(accuracy)
```

#### statistics
Statistics is the branch of mathematics that deals with collecting, analyzing, interpreting, and presenting data.   

                Why Do We Need Statistics? 
   
* Summarize large amounts of data.    
* Identify patterns and trends.   
* Collect data systematically.    
* Compare different groups.   
* Make predictions and informed decisions.    

        Example 

Suppose a teacher wants to know how students performed in an exam.      

The marks are:  
70, 75, 80, 85, 90      

Using statistics, the teacher can:      

* Find the average (mean) score.
* Identify the highest and lowest marks.
* Measure how spread out the scores are.
* Compare this class with another class.




#### branches of statistics
- collection of data
- discription

- analysis

        collection of data    
Data collection is the systematic process of gathering, measuring, and recording information from various sources for analysis and decision-making. 


        discription    
Description is a detailed explanation that tells what something is, what it is like, and its important features.    

        Presentation of data    
Presentation of data is the process of organizing and displaying collected data in a clear, systematic, and meaningful form so that it can be easily understood, analyzed, and interpreted. 

        analysis    
Analysis of data is the process of examining, organizing, summarizing, and interpreting collected data to discover useful information, identify patterns, and support decision-making.

## Types of Statistics
1. Descriptive Statistics
2. Inferential Statistics

### Descriptive Statistics
Descriptive Statistics is the branch of statistics that collects, organizes, summarizes, and presents data in a meaningful way. It helps us understand the main characteristics of a dataset without making predictions or drawing conclusions about a larger population.   

Descriptive Statistics is the process of organizing, summarizing, and presenting data using numerical measures, tables, and graphs. 

* Example
Suppose the marks of 10 students are:   
60, 65, 70, 75, 80, 85, 90, 75, 70, 80  
- Instead of looking at all 10 values individually, descriptive statistics summarizes the data by calculating:    
- Average (Mean)
- Middle value (Median)
- Most frequent value (Mode)
- Spread of data (Range, Variance, Standard Deviation)  

                Objectives of Descriptive Statistics
* Organize raw data
* Summarize large datasets
* Identify patterns
* Compare datasets
* Present information using tables and graphs
* Understand the distribution of data

            Types of Descriptive Statistics
1. Measures of Central Tendency
These describe the center of the dataset.   
- Mean (Average)       
The mean is the sum of all observations divided by the number of observations   

- Median        
The median is the middle value after arranging data in order.   

- Mode    
The mode is the value that appears most frequently.     

2. Measures of Dispersion   
These measure how spread out the data is.   

- Range   
Range = Maximum − Minimum   

- Variance   
Variance measures how far the data points are from the mean on average (using squared deviations).  

- Standard Deviation    
Standard deviation is the square root of the variance. It measures the typical distance of data values from the mean.   

            Data Presentation Methods
Descriptive statistics also presents data visually. 
##### Tables
* Frequency table
* Contingency table
##### Graphs
* Bar chart
* Histogram
* Pie chart
* Line graph
* Box plot
* Scatter plot

### Inferential Statistics

Inferential Statistics is the branch of statistics that uses sample data to make predictions, estimates, or conclusions about an entire population. 

Inferential statistics is the process of drawing conclusions about a population based on information collected from a sample.   


            Purpose of Inferential Statistics
The purpose of inferential statistics is to use sample data to make reliable conclusions, predictions, or decisions about a larger population without collecting data from every individual.    

        Main Purposes
1. To draw conclusions about a population
* Uses information from a sample to estimate characteristics of the entire population.
2. To make predictions
* Predicts future outcomes or trends based on sample data.
3. To estimate population parameters
* Estimates values such as the population mean, proportion, or variance.
4. To test hypotheses
* Determines whether a claim or assumption about a population is supported by sample data.
5. To support decision-making
* Helps businesses, governments, and researchers make informed decisions based on statistical evidence.

            Characteristics of Inferential Statistics
The following are the main characteristics of inferential statistics:   

1. Based on Sample Data
* Inferential statistics uses data collected from a sample rather than the entire population.
2. Generalizes to the Population
* It draws conclusions or makes predictions about the whole population based on sample results.
3. Uses Probability Theory
* It relies on probability to measure uncertainty and determine how likely conclusions are to be correct.
4. Estimates Population Parameters
* It estimates unknown population values such as the population mean, proportion, and variance.
5. Tests Hypotheses
* It evaluates claims or assumptions about a population using statistical tests.
