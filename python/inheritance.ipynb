### Single Inheritance :-
When one child class inherits from one parent class.

```python
class Father:
    pass
class Son(Father):
    pass
```

```python
class Father:
    father_name="jai"
    father_account="12342234"
class son(Father):
    son_name="rohit"
beta=son()
print(beta.father_name)
print(beta.son_name)
print(beta.father_account)
```

### Multiple Inheritance :- 
When one child class inherits from multiple parent classes.

```python
class GrandFather:
    pass
class Father(GrandFather):
    pass
class son(Father):
    pass
```

```python
class GrandFather:
    grand_father_account="652983"
class Father(GrandFather):
    father_account="6387994"
class son(Father):
    print("mai chill krunga")

beta=son()

print(beta.grand_father_account)
print(beta.father_account)
```

### Multilevel Inheritance :- 
When a child class becomes a parent for another class.

```python
class Employes:
    def company_info(self):
        print("company name : techsoft")

class Developer(Employes):
    def developer_work(self):
        print("developer write and test code")

class TeamLead(Developer):
    def manage_team(self):
        print("team lead and manage the developer")

obj=TeamLead()

obj.company_info()
obj.developer_work()
obj.manage_team()


```

```python
# mutiple inheritance syntax

class Father:
    pass
class Uncle:
    pass
class son(Father,Uncle):
    pass
```

```python
class Father:
    father_ammount=745784
class Uncle:
    uncle_ammount=638835
class son(Father,Uncle):
    son_ammount=0

obj=son()

print(obj.father_ammount)
print(obj.uncle_ammount)
print(obj.son_ammount)
```

```python
class Employes:
    def company_info(self):
        print("company name : techsoft pvt lmt")

class Developer:
    def developer_work(self):
        print("developer write and test code")

class TeamLead:
    def manage_team(self):
        print("team lead and manage the developer")

class Founder(Employes,Developer,TeamLead):
    def work_mistake(self):
        print("no problem")

obj=Founder()

obj.company_info()
obj.developer_work()
obj.manage_team()
obj.work_mistake()
```

### hybrid inheritance :- 
When different inheritance types are combined in one program.

```python

class GrandFather:
    def grandfather_ammount(self):
        print("ammount : 3665")

class Father(GrandFather):
    def father_ammount(self):
        print("father ammount : 63563")

class Uncle(GrandFather):
    def uncle_ammount(self):
        print("uncle ammount : 63774")

class Brother(Father,Uncle):
    def brother_ammount(self):
        print("brother ammount : 73564")

obj=Brother()

obj.grandfather_ammount()
obj.father_ammount()
obj.uncle_ammount()
obj.brother_ammount()
```

### Hierarchical Inheritance :-
When many child classes inherit from one parent class.

```python

class Father:
    def father_ammount(self):
        print("father ammount : 6374")
class Child1(Father):
    def child1_ammount(self):
        print("child1 ammount : 635")
class Child2(Father):
    def child2_ammount(self):
        print("child2 ammount : 432")

obj1=Child1()
obj2=Child2()

obj1.father_ammount()
obj1.child1_ammount()

obj2.father_ammount()
obj2.child2_ammount()
```
