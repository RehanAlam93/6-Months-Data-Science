### STRING  IN PYTHON  :-
A string in Python is a data type used to store a sequence of characters (text). It can include letters, numbers, symbols, and spaces. Strings are written inside quotes: single (' '), double (" "), or triple quotes (''' ''' / """ """).

```python
name = "Rehan"
print(name)
```

### indexing in pyhon :- 
Indexing means accessing a single character from a string using its position number (index).
In Python, indexing starts from 0 for the first character and can also use negative indexing from the end.

```python
text = "Python"
print(text[0])   
print(text[3])   
print(text[-1])
```

### slicing in python :-
Slicing in Python is the process of extracting a part (sub-string) of a string by using a range of index values.
It allows you to get multiple characters from a string at once using the format:

string[start : end : step]

```python
print(text[0:4])                  
print(text[2:])                    
print(text[:3])
```

### Set Operators in Python
Python sets support different operators to perform operations like union, intersection, difference, etc.

* union(|) :-  Definition: Combines all elements from both sets (duplicates removed) 
* Intersection (&) :- Definition: Gives common elements from both sets
* Difference (-) :- Definition: Elements present in first set but not in second
* Symmetric Difference (^) :- Definition: Elements present in either set, but not in both

```python
a = {1, 2, 3}
b = {3, 4, 5}
print("union :- ",a | b)
print("intersection :- ",a & b)
print("difference :- ",a - b)
print("Symmetric Difference :- ",a ^ b)
```

### Inheritance in Python :-
Inheritance in Python is a feature of Object-Oriented Programming (OOP) in which one class acquires the properties and methods of another class.

                    Types of Inheritance in Python
* Single Inheritance
* Multiple Inheritance
* Multilevel Inheritance
* Hierarchical Inheritance
* Hybrid Inheritance

#### Single Inheritance :-
One child class inherits from one parent class.

```python
class Parent:
    def show(self):
        print("This is parent class")

class Child(Parent):
    def display(self):
        print("This is child class")

obj = Child()

obj.show()
obj.display()
```

#### Multiple Inheritance :-
One child class inherits from more than one parent class.

```python
class Father:
    def skill1(self):
        print("Driving")

class Mother:
    def skill2(self):
        print("Cooking")

class Child(Father, Mother):
    pass

obj = Child()

obj.skill1()
obj.skill2()
```

#### Multilevel Inheritance :-
A child class becomes the parent for another class.

```python
class Grandfather:
    def house(self):
        print("House from Grandfather")

class Father(Grandfather):
    pass

class Son(Father):
    pass

obj = Son()

obj.house()
```

#### Hierarchical Inheritance :-
Multiple child classes inherit from the same parent class.

```python
class Parent:
    def show(self):
        print("Parent class")

class Child1(Parent):
    pass

class Child2(Parent):
    pass

obj1 = Child1()
obj2 = Child2()

obj1.show()
obj2.show()
```

#### Hybrid Inheritance :-
Combination of more than one type of inheritance.

```python
class A:
    def methodA(self):
        print("Class A")

class B(A):
    pass

class C(A):
    pass

class D(B, C):
    pass

obj = D()

obj.methodA()
```
