```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#### Matplotlib :- 
Matplotlib is a popular Python library used for creating data visualizations such as line charts, bar charts, pie charts, scatter plots, histograms, and more. It helps present data in a graphical form, making it easier to analyze and understand.

                Features
* Creates a wide variety of plots and charts.
* Easy to use with NumPy and pandas.
* Highly customizable (titles, labels, colors, legends, etc.).
* Suitable for both simple and advanced visualizations.

        The code displays a line graph where:
* The X-axis contains the values 1, 2, 3, 4, 5.
* The Y-axis contains the values 10, 20, 15, 30, 25.
* The line connects the data points to show the trend.

        Other Common Plots in Matplotlib
* Line Plot – Shows trends over time or ordered data.
* Bar Chart – Compares values across categories.
* Scatter Plot – Shows the relationship between two variables.
* Histogram – Displays the distribution of numerical data.
* Pie Chart – Shows proportions or percentages of a whole.

```python
ypoint=np.array([2,4,6,8])
xpoint=np.array([1,5,7,9])
plt.plot(ypoint)
plt.plot(xpoint)
plt.show()
```

```python
ypoint=np.array([2,4,6,8])
xpoint=np.array([1,5,7,9])
plt.ylabel("wcss")
plt.xlabel("weight")
plt.title("Elbow method")
plt.plot(ypoint,xpoint)
plt.show()
```

```python
ypoint=np.array([2,4,6,8])
xpoint=np.array([1,5,7,9])
plt.bar(ypoint,xpoint, color=['red',"blue","green","black"])
```

```python
ypoint=np.array([2,4,6,8])
xpoint=np.array([1,5,7,9])
plt.ylabel("wcss")
plt.xlabel("weight")
plt.title("Elbow method")

plt.bar(ypoint,xpoint, color="red",width=0.5)
```
