```python
class College:
    def student(self,student_behaviour):
        self.student_behaviour=student_behaviour
class Teacher(College):
    def teacher(self,techer_knowledge,teach):
        self.techer_knowledge=techer_knowledge
        self.teach=teach

obj=Teacher()

obj.student("student behaviour is good")
obj.teacher("teacher is qulified","teacher teaches well")

print(obj.student_behaviour)
print(obj.techer_knowledge)
print(obj.teach)
```
