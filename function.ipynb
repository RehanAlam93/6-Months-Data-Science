### What is a Function?
A function in Python is a block of reusable code that performs a specific task.
It helps make programs more organized, readable, and reusable.

* Avoid repeating code
* Improve readability
* Make debugging easier
* Break large problems into smaller parts

        Types of Functions in Python

#### Built-in Functions
These are functions already provided by Python. You don’t need to define them.

```python
print("Hello")
len("Python")
type(10)
int("5")
```

#### User-defined Functions
These are functions created by the programmer using def.


```python
def greet(name):
    return "Hello " + name

print(greet("Rehan"))
```

```python
lst=[1,2,3,4,5,6]
def even_odd(lst):
    print("this is my list : ",lst)
    for x in lst:
        if x%2==0:
            print("even",x)
        else:
            print("odd",x)
even_odd(lst=lst)
```

```python
num =5
def fact(num):
    a = 1
    while num >= 1:
        a = a*num
        num-= 1
    return a
print(fact(num))
```

```python
def call_list(lst):
    total=0
    for x in lst:
        total+=x
    return total
call_list(lst=[1,2,3,4,5,6,7])
        
```

### method :-  class ke andar jo fenction pass kiya jata hai usko method bolte hai
a **method** is a function that belongs to an object (or a class). Methods are used to define the behavior of objects.
A **method** is a function defined inside a class and is called using an object of that class

### function :-  
A function is a block of code that performs a specific task. It helps avoid repeating code and makes programs easier to understand and manage.





```python
def greet():     # function example
    print("Hello, Welcome!")

greet()  # Calling the function
```

### Method in Python  :-
A method is a function that is defined inside a class and is used to perform actions related to objects of that class.

* example

```python
class Car:
    # Method
    def start(self):
        print("Car is starting")

# Create object
my_car = Car()

# Call method
my_car.start()
```

### Parameter in Python :-
A parameter is a variable written inside the parentheses of a function definition. It is used to receive values when the function is called.
* example


```python
def greet(name):   # 'name' is a parameter
    print("Hello", name)

greet("Rehan")     # "Rehan" is an argument
```

### object :- 
An object is an instance of a class. It is created from a class and is used to access the properties and methods of that class.
* Example

```python
class Student:
    def greet(self):
        print("Hello Student")

# Create object
s1 = Student()

# Access method using object
s1.greet()
```
