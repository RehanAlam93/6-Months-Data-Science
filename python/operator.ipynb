## Arithmetic Operators
| Operator | Name                | Example   | Output |
| -------- | ------------------- | --------- | ------ |
| `+`      | Addition            | `10 + 5`  | `15`   |
| `-`      | Subtraction         | `10 - 5`  | `5`    |
| `*`      | Multiplication      | `10 * 5`  | `50`   |
| `/`      | Division            | `10 / 5`  | `2.0`  |
| `%`      | Modulus (Remainder) | `10 % 3`  | `1`    |
| `**`     | Exponent (Power)    | `2 ** 3`  | `8`    |
| `//`     | Floor Division      | `10 // 3` | `3`    |

### Bitwise Operators

| -------- | ----------- | ---------- | -------- | --- |

| `~`      | Bitwise NOT | `~5`       | `-6`     |     |
| `<<`     | Left Shift  | `5 << 1`   | `10`     |     |
| `>>`     | Right Shift | `5 >> 1`   | `2`      |     |
