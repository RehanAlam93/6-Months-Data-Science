#### Matplotlib :- 
Matplotlib is a popular Python library used for creating data visualizations such as line charts, bar charts, pie charts, scatter plots, histograms, and more. It helps present data in a graphical form, making it easier to analyze and understand.

                Features
* Creates a wide variety of plots and charts.
* Easy to use with NumPy and pandas.
* Highly customizable (titles, labels, colors, legends, etc.).
* Suitable for both simple and advanced visualizations.

        The code displays a line graph where:
* The X-axis contains the values 1, 2, 3, 4, 5.
* The Y-axis contains the values 10, 20, 15, 30, 25.
* The line connects the data points to show the trend.

        Other Common Plots in Matplotlib
* Line Plot – Shows trends over time or ordered data.
* Bar Chart – Compares values across categories.
* Scatter Plot – Shows the relationship between two variables.
* Histogram – Displays the distribution of numerical data.
* Pie Chart – Shows proportions or percentages of a whole.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load the dataset
# Replace 'Used_Bikes.csv' with your actual file path if it's stored elsewhere
df=pd.read_csv(r"C:\Users\rehan\OneDrive\Desktop\Data Science\6-Months-Data-Science\pandas\Used_Bikes.csv")


# 2. Data Cleaning: Drop exact duplicates to avoid skewed visualizations
df = df.drop_duplicates()

# 3. Filter for the Top 5 most frequent brands for a cleaner plot
top_brands = df["brand"].value_counts().nlargest(5).index
df_filtered = df[df["brand"].isin(top_brands)]

# 4. Set up the matplotlib figure
plt.figure(figsize=(12, 7))
sns.set_theme(style="whitegrid")

# 5. Create a scatter plot with a trend line (lmplot equivalent via regplot/scatterplot)
# We limit the price to 300,000 to focus on the bulk of the data
sns.scatterplot(
    data=df_filtered,
    x="age",
    y="price",
    hue="brand",
    alpha=0.6,
    palette="deep",
    s=50,
)

# 6. Customize the chart labels and formatting
plt.title(
    "Impact of Bike Age on Price (Top 5 Brands)", fontsize=16, fontweight="bold"
)
plt.xlabel("Age of the Bike (Years)", fontsize=12)
plt.ylabel("Price (INR)", fontsize=12)
plt.xlim(0, 15)  # Focus on bikes under 15 years old
plt.ylim(0, 300000)  # Focus on the most common price range

# Format the y-axis with commas for better readability
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"{int(x):,}"))

# Adjust legend position so it doesn't block data points
plt.legend(title="Brand", bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()

# 7. Display the graph
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load the dataset
df = pd.read_csv(r"C:\Users\rehan\OneDrive\Desktop\Data Science\6-Months-Data-Science\pandas\Used_Bikes.csv")
df = df.drop_duplicates()

# 2. Set up the matplotlib figure
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# 3. Create a boxplot of Price vs Owner Type
# 'showfliers=False' keeps the focus on the main distribution by removing extreme outliers
sns.boxplot(
    data=df,
    x="owner",
    y="price",
    showfliers=False,
    palette="Set2",
    order=["First Owner", "Second Owner", "Third Owner", "Fourth Owner Or More"],
)

# 4. Customize labels and formatting
plt.title(
    "Used Bike Price Distribution by Owner Type", fontsize=15, fontweight="bold"
)
plt.xlabel("Owner Type", fontsize=12)
plt.ylabel("Price (INR)", fontsize=12)

# Format y-axis with commas for readability
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"{int(x):,}"))

# Rotate x-axis labels slightly so they don't overlap
plt.xticks(rotation=15)

plt.tight_layout()

# 5. Display the graph
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load the dataset and remove duplicates
df = pd.read_csv(r"C:\Users\rehan\OneDrive\Desktop\Data Science\6-Months-Data-Science\pandas\Used_Bikes.csv")

df = df.drop_duplicates()

# 2. Filter out rare brands (less than 5 bikes) to ensure clean data
brand_counts = df["brand"].value_counts()
top_brands = brand_counts[brand_counts >= 5].index
df_filtered = df[df["brand"].isin(top_brands)]

# 3. Calculate the median price per brand to order the graph beautifully
brand_order = (
    df_filtered.groupby("brand")["price"]
    .median()
    .sort_values(ascending=False)
    .index
)

# 4. Set up the matplotlib figure
plt.figure(figsize=(14, 8))
sns.set_theme(style="whitegrid")

# 5. Create a horizontal bar plot
sns.barplot(
    data=df_filtered,
    x="price",
    y="brand",
    order=brand_order,
    estimator="median",  # Focuses on the typical price point
    errorbar=None,  # Removes confidence intervals for a cleaner look
    palette="viridis",
)

# 6. Customize labels and formatting
plt.title(
    "Typical (Median) Price by Bike Brand", fontsize=16, fontweight="bold", pad=15
)
plt.xlabel("Median Price (INR)", fontsize=12)
plt.ylabel("Brand", fontsize=12)

# Format the x-axis with commas for better readability
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"{int(x):,}"))

plt.tight_layout()

# 7. Display the graph
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load the dataset and remove duplicates
df = pd.read_csv(r"C:\Users\rehan\OneDrive\Desktop\Data Science\6-Months-Data-Science\pandas\Used_Bikes.csv")

df = df.drop_duplicates()

# 2. Filter data to focus on the main market distribution (helps remove extreme outliers)
df_filtered = df[(df["kms_driven"] <= 100000) & (df["price"] <= 300000)]

# 3. Set up the matplotlib figure
plt.figure(figsize=(11, 6))
sns.set_theme(style="whitegrid")

# 4. Create a scatter plot with low alpha (transparency) to manage overlapping data points
sns.scatterplot(
    data=df_filtered,
    x="kms_driven",
    y="price",
    alpha=0.3,
    color="teal",
    s=30,
)

# 5. Add a linear trendline to visualize the depreciation direction
sns.regplot(
    data=df_filtered,
    x="kms_driven",
    y="price",
    scatter=False,
    color="red",
    line_kws={"linewidth": 2, "linestyle": "--"},
)

# 6. Customize labels and number formatting
plt.title("Bike Price vs. Kilometers Driven", fontsize=15, fontweight="bold", pad=15)
plt.xlabel("Kilometers Driven (kms)", fontsize=12)
plt.ylabel("Price (INR)", fontsize=12)

# Format axes with commas for clean readability
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"{int(x):,}"))
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"{int(x):,}"))

plt.tight_layout()

# 7. Display the graph
plt.show()
```
