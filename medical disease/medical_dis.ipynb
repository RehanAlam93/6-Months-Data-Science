```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
```

```python
df=pd.read_csv('medical_disease.csv')   # read medical_disease.csv file and store in dataframe df
```

```python
df
```

```python
print("duplicate values : ",df.duplicated().sum())
print("null values : ",df.isnull().sum())
```

```python
df['Gender'].unique()
```

```python
df['Gender']=df['Gender'].map({'Male':1,'Female':0}) # map the values 
```

```python
df['Smoking'].unique()
```

```python
df['Smoking'] = df['Smoking'].str.strip().str.title().map({'No':0, 'Yes':1}) # map the values of Smoking column to 0 and 1 and also remove any leading or trailing spaces and convert to title case
```

```python
df['Alcohol'].unique()
```

```python
df['Alcohol'] = df['Alcohol'].str.strip().str.title().map({'No':0, 'Yes':1})
```

```python
df['PhysicalActivity'].unique()
```

```python
df['PhysicalActivity']=df['PhysicalActivity'].str.strip().str.title().map({'Medium':2,'High':3,'Low':1})
```

```python
df['FamilyHistory'].unique()
```

```python
df['FamilyHistory']=df['FamilyHistory'].str.strip().str.title().map({'No':0,'Yes':1})
```

```python
df['Disease'].unique()
```

```python
df['Disease']=df['Disease'].str.strip().str.title().map({'Healthy':0,'Pre-Diabetes':1,'Hypertension':2,'Heart Disease':3,'Diabetes':4})
```
