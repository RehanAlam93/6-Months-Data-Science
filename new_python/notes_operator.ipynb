### operator in python  
An operator is a symbol that is used to perform a specific action on values.

        Types of operator in python
* Arithmetic operators
* Comparison operators
* Logical operators
* Assignment operators
* Bitwise operators
* Membership operators
* Identity operators



### Arithmetic operators :- 
Arithmetic operators are operators that are used to perform mathematical operations like addition, subtraction, multiplication, division, etc.

| Operator | Name           | Description                                   | Example |
|----------|----------------|-----------------------------------------------|---------|
| +        | Addition       | Adds together two values                      | x + y   |
| -        | Subtraction    | Subtracts one value from another              | x - y   |
| *        | Multiplication | Multiplies two values                         | x * y   |
| /        | Division       | Divides one value by another                  | x / y   |
| %        | Modulus        | Returns the division remainder                | x % y   |
| ++       | Increment      | Increases the value of a variable by 1        | ++x     |
| --       | Decrement      | Decreases the value of a variable by 1        | --x     |



```python
x = 10
y = 4
print("addition : ",x+y)
print("subtraction : ",x-y)
print("multiplication : ",x*y)
print("Division of : ",x/y)
print("Modulus of : ",x%y)
```

### Comparison operators :-  
Comparison operators are used to compare two values. They always return a result in True or False (Boolean value).

| Operator | Name                  | Description                          | Example |
|----------|-----------------------|--------------------------------------|---------|
| ==       | Equal to              | Checks if two values are equal       | x == y  |
| !=       | Not equal to          | Checks if two values are not equal   | x != y  |
| >        | Greater than          | Checks if left value is greater      | x > y   |
| <        | Less than             | Checks if left value is smaller      | x < y   |
| >=       | Greater than or equal | Checks if left value is greater/equal| x >= y  |
| <=       | Less than or equal    | Checks if left value is smaller/equal| x <= y  |

```python
a = 15
b = 10
print("equal to : ",a==b)
print("not equal to : ",a!=b)
print("a is greater than b : ",a>b)
print("a is greater than or equal to b : ",a>=b)

```

### Logical operators  :-
Logical operators are operators that are used to combine two or more conditions and return True or False.

| Operator   | Name     | Description                                    | Example            |
|------------|----------|------------------------------------------------|--------------------|
| and        | AND      | Returns True if both conditions are True       | x > 5 and y < 10   |
| or         | OR       | Returns True if at least one condition is True | x > 5 or y < 10    |
| not        | NOT      |Reverses the result (True to False)             | not(x > 5)         |


```python
x = 10
y = 5
print("AND operator : ",x>50 and y<2)
print("OR Operator : ",x>5 or y>20)
print("NOT operator : ",not(x>5))
```

### Assignment operators :- 
Assignment operators are operators that are used to assign values to variables.

| Operator | Meaning                  | Example   |
|----------|--------------------------|-----------|
| `=`      | Assign value             | `x = 5`   |
| `+=`     | Add and assign           | `x += 5`  |
| `-=`     | Subtract and assign      | `x -= 5`  |
| `*=`     | Multiply and assign      | `x *= 5`  |
| `/=`     | Divide and assign        | `x /= 5`  |
| `%=`     | Modulus and assign       | `x %= 5`  |
| `//=`    | Floor divide and assign  | `x //= 5` |
| `**=`    | Power and assign         | `x **= 5` |


```python
#assign value
x= 8
print("assign value of x : ",x)
# add and assign
x+=5
print("After add : ",x)
# subtract and assign
x-=3
print("After subtract : ",x)
# multiply and assign
x*=5
print("after multiply : ",x)
# Divide and assign
x/=5
print("after Divide : ",x)


```

### Bitwise Operators :- 
Bitwise operators work on binary numbers (0 and 1) and perform operations bit by bit.

| Operator  | Name        | Meaning                  | Example    |
|-----------|-------------|--------------------------|------------|
| `&`       | AND         | Both bits must be 1      | `x & y`    |
| `\|`      | OR          | At least one bit is 1    | `x \| y`   |
| `^`       | XOR         | Different bits return 1  | `x ^ y`    |
| `~`       | NOT         | Reverse the bits         | `~x`       |
| `<<`      | Left Shift  | Shift bits to left       | `x << 2`   |
| `>>`      | Right Shift | Shift bits to right      | `x >> 2`   |


```python
x = 5
y = 3
print("AND :", x & y)
print("Or :", x | y)
print("XOR :", x ^ y)
print("NOT :", ~x)


```

### Membership Operators :- 
Membership operators are used to check whether a value is present in a sequence such as list, tuple, string, or dictionary.
(Membership operators are used to check if an element exists in a sequence or not.)

| Operator  | Meaning                                |
|-----------|----------------------------------------|
| `in`      | Returns `True` if value is present     |
| `not in`  | Returns `True` if value is not present |


```python
fruits = ["apple", "banana", "mango"]
print("apple" in fruits)
print("orange" not in fruits)
print("grapes" in fruits)
print("mango" not in fruits)
```

### Identity Operators :- 
Identity operators are used to compare the memory location of two objects. They check whether two variables refer to the same object or not.




| Operator | Meaning                                                          |
|-----------|-----------------------------------------------------------------|
| `is` | Returns `True` if both variables refer to the same object            |
| `is not` | Returns `True` if both variables do not refer to the same object |

```python
x = 10
y = 10
z = 20

print(x is y)
print(x is z)
print(x is not z)
```
