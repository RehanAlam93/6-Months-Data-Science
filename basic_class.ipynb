* public variable

```python
class College:
    collage_library="yes"     # public
    account_info=645373
    college_name="CIT"
cr=College()
print(cr.collage_library)
print(cr.account_info)
print(cr.college_name)
```

```python
class College:
    collage_library="yes"     
    _account_info=645373   # private   ( by use single underscore _)
    college_name="CIT"
cr=College()
cr.collage_library
cr.account_info
cr.college_name
```

```python
class College:
    collage_library="yes"    
    __account_info=645373     # protected  (by use doble underscore __ )
    college_name="CIT"
cr=College()
cr.collage_library
cr.account_info
cr.college_name
```

```python
class Company:
    _account_details=74664645
    def account(self):
        return self._account_details
cr=Company()
cr._account_details
```
