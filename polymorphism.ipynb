```python
class School:
    def init(self,name,branch):
        self.name=name
        self.branch=branch
class Private(School):
    def private(self,roll_number):
        super().init("Rehan","CSE")
        self._roll_number=roll_number

obj=Private()

obj.private(42)
```

```python
print(obj.name)
print(obj.branch)
print(obj._roll_number)
```

```python
class Public:
    def info_1(self,company_name,employe_name):
        self.company_name=company_name
        self.employe_name=employe_name

class Protected:
    def info_2(self,work,contact):
        self._work=work
        self._contact=contact

class Private(Public,Protected):
    def info_3(self,turnover):
        self.__turnover=turnover

obj=Private()

obj.info_1("InfoTech","Rehan")
obj.info_2("Developer",9334993095)
obj.info_3("$80 million")

        
        
```

```python
print(obj.company_name)
print(obj.employe_name)
print(obj._work)
print(obj._contact)
print(obj._Private__turnover)
```
