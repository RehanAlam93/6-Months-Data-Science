```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=sns.load_dataset('tips')
```

```python
data
```

```python
sns.histplot(data['total_bill'])
plt.show()
```

```python
sns.scatterplot(x=data['total_bill'], y=data['tip'])
plt.show()
```

```python
plt.scatter(data['tip'],data['total_bill'])
plt.show()
```

```python
data
```

```python
sns.kdeplot(data=data,x='tip',y='total_bill')
plt.show()
```

```python
sns.kdeplot(data=data,x='total_bill')
plt.show()
```

```python
sns.relplot(x='tip',y='total_bill',data=data)
plt.show()
```

```python
sns.relplot(x='tip',y='total_bill',data=data,hue='smoker',style='time')
plt.show()
```

```python
sns.relplot(x='tip',y='total_bill',data=data,hue='smoker',col='day')
plt.show()
```

```python
sns.catplot(x=data['day'],y=data['total_bill'])
plt.show()
```

```python
sns.catplot(x=data['day'],y=data['total_bill'],kind='violin')
plt.show()
```

```python
sns.catplot(x=data['day'],y=data['total_bill'],kind='box')
plt.show()
```

```python
sns.catplot(x=data['day'],y=data['total_bill'],kind='boxen')
plt.show()
```

```python
sns.catplot(x=data['day'],y=data['total_bill'],kind='strip')
plt.show()
```

```python
sns.catplot(x=data['day'],y=data['total_bill'],kind='swarm')
plt.show()
```
